```python
#outlier handling in sqft_lot
#Outlair handling using iqr method
q1 = house_data['sqft_lot'].quantile(0.25)
q2 = house_data['sqft_lot'].quantile(0.50)
q3 = house_data['sqft_lot'].quantile(0.75)

print(f'Q1 = {q1}\nQ2 = {q2}\nQ3 = {q3}')

iqr = q3 - q1

up_lim = q3 + (1.5 * iqr)
low_lim = q1 - (1.5 * iqr)

print(f'IQR = {iqr}\nUp_lim = {up_lim}\nLow_lim = {low_lim}')
```

replacing outlier with upper limit or lower limit

```python
outliers = house_data[(house_data['sqft_lot'] > up_lim) | (house_data['sqft_lot'] < low_lim)]
outliers
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
model = lr.fit(x_train1, y_train1)
y_pred = model.predict(x_test1)
#using MSR, r2_score matrices for model evaluation
from sklearn.metrics import mean_squared_error,r2_score,mean_absolute_error
print(mean_squared_error(y_test1, y_pred))
print(r2_score(y_test1, y_pred))
print(mean_absolute_error(y_test1, y_pred))
```

#Lasso Regration

```python
from sklearn.linear_model import Lasso
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

lasso = Lasso(alpha=10,max_iter=10000)
lasso_model = lasso.fit(x_train1, y_train1)
y_pred_lasso = lasso_model.predict(x_test1)
mse = mean_squared_error(y_test1, y_pred_lasso)
r2 = r2_score(y_test1, y_pred_lasso)
mae = mean_absolute_error(y_test1, y_pred_lasso)

print(f"Lasso Regression MSE: {mse}")
print(f"Lasso Regression R2 Score: {r2}")
print(f"Lasso Regression MAE: {mae}")
```
